```python

# CELL 7 - HOME CREDIT DATA INGESTION

import hashlib
import os

# Define paths
source_path = "/Volumes/dbw_homecredit/homecredit_bronze/homecredit_uploads"
version_path = f"{source_path}/homecredit_raw/v1"

os.makedirs(version_path, exist_ok=True)

# Function to calculate SHA-256 hash
def calculate_hash(path):
    if path.startswith("dbfs:/Volumes/"):
        path = path.replace("dbfs:", "", 1)

    sha256 = hashlib.sha256()

    with open(path, "rb") as file:
        for chunk in iter(
            lambda: file.read(8 * 1024 * 1024),
            b""
        ):
            sha256.update(chunk)

    return sha256.hexdigest()


# Find all CSV files
source_files = sorted(
    [
        f for f in os.scandir(source_path)
        if f.is_file() and f.name.lower().endswith(".csv")
    ],
    key=lambda f: f.name
)

manifest_files = []

# Copy and verify each file
for file in source_files:

    source = file.path
    destination = os.path.join(version_path, file.name)

    source_hash = calculate_hash(source)
    source_size = os.path.getsize(source)

    if os.path.exists(destination):

        destination_hash = calculate_hash(destination)

        if (
            source_hash == destination_hash
            and os.path.getsize(destination) == source_size
        ):
            print("SKIPPED (identical):", file.name)
        else:
            raise ValueError(
                f"Hash mismatch: {file.name}. "
                "Existing file differs from source."
            )

    else:
        # Copy file in chunks
        temporary_path = destination + ".partial"

        try:
            with open(source, "rb") as src:
                with open(temporary_path, "wb") as dst:
                    while True:
                        chunk = src.read(8 * 1024 * 1024)
                        if not chunk:
                            break
                        dst.write(chunk)

            # Verify copied file
            copied_hash = calculate_hash(temporary_path)

            if copied_hash != source_hash:
                raise ValueError(
                    f"Copy verification failed: {file.name}"
                )

            os.replace(temporary_path, destination)

        finally:
            if os.path.exists(temporary_path):
                os.remove(temporary_path)

        print("COPIED:", file.name)

    # Add file information to manifest list
    manifest_files.append({
        "filename": file.name,
        "size_bytes": source_size,
        "sha256": source_hash
    })

print("--------------------------------")
print("Processed files:", len(manifest_files))
print("Cell 7 completed successfully!")
```

```python

# CELL 8 - GENERATE MANIFEST.JSON

import json
import os
from datetime import datetime, timezone

manifest_path = os.path.join(version_path, "manifest.json")

manifest = {
    "project": "Home Credit Default Risk",
    "version": "v1",
    "created_at": datetime.now(timezone.utc).isoformat(),
    "file_count": len(manifest_files),
    "files": manifest_files
}

if os.path.exists(manifest_path):

    with open(manifest_path, "r", encoding="utf-8") as f:
        existing_manifest = json.load(f)

    def file_records(items):
        return {
            item["filename"]: (
                item["size_bytes"],
                item["sha256"]
            )
            for item in items
        }

    if (
        existing_manifest.get("version") != "v1"
        or existing_manifest.get("file_count") != len(manifest_files)
        or file_records(existing_manifest["files"])
        != file_records(manifest_files)
    ):
        raise ValueError(
            "Existing manifest differs from current dataset."
        )

    print("SKIPPED: Existing manifest verified.")

else:
    # Write manifest without overwriting an existing one
    with open(manifest_path, "x", encoding="utf-8") as f:
        json.dump(manifest, f, indent=4)

    print("SUCCESS: manifest.json created!")

print("Manifest location:", manifest_path)
print("Total CSV files:", len(manifest_files))
```

```python

# CELL 9 - VERIFY MANIFEST

with open(manifest_path, "r", encoding="utf-8") as f:
    saved_manifest = json.load(f)

print("Project:", saved_manifest["project"])
print("Version:", saved_manifest["version"])
print("Total files:", saved_manifest["file_count"])

for file in saved_manifest["files"]:
    print(
        file["filename"],
        "| SHA-256:",
        file["sha256"][:16] + "..."
    )

print("\nManifest verification completed!")
```

```python

# CELL 10 - IDEMPOTENCY VERIFICATION

import os
import json

with open(manifest_path, "r", encoding="utf-8") as f:
    manifest = json.load(f)

expected_files = {
    item["filename"]: item
    for item in manifest["files"]
}

actual_files = {
    f.name: f.path
    for f in os.scandir(version_path)
    if f.is_file() and f.name.lower().endswith(".csv")
}

assert set(expected_files) == set(actual_files), (
    "File list does not match manifest!"
)

for filename, metadata in expected_files.items():
    actual_path = actual_files[filename]

    assert os.path.getsize(actual_path) == metadata["size_bytes"], (
        f"Size mismatch: {filename}"
    )

    assert calculate_hash(actual_path) == metadata["sha256"], (
        f"SHA-256 mismatch: {filename}"
    )

print("SUCCESS: All files match the manifest.")
print("Verified files:", len(expected_files))
print("No duplicate CSV filenames detected.")
print("Idempotency verification passed.")
```

```python

# CELL 11 - CONTROLLED FAILURE TEST (CORRECTED)

import os
import hashlib

source_path = "/Volumes/dbw_homecredit/homecredit_bronze/homecredit_uploads"

test_path = f"{source_path}/failure_test"

dbutils.fs.mkdirs(test_path)

# Create a small, valid sample file
original_file = f"{test_path}/sample_valid.csv"
test_file = f"{test_path}/sample_corrupted.csv"

valid_data = (
    "SK_ID_CURR,TARGET,AMT_INCOME_TOTAL\n"
    "100001,0,202500\n"
    "100002,1,135000\n"
)

corrupted_data = (
    "SK_ID_CURR,TARGET,AMT_INCOME_TOTAL\n"
    "100001,0,202500\n"
    "100002,INVALID,135000\n"
)

# Write the files using Databricks utilities
dbutils.fs.put(original_file, valid_data, overwrite=True)
dbutils.fs.put(test_file, corrupted_data, overwrite=True)

print("Valid sample created:", original_file)
print("Corrupted sample created:", test_file)
print("Original Home Credit dataset remains unchanged.")
```

```python

# CELL 12 - DETECT CORRUPTED DATA

original_hash = calculate_hash(original_file)
test_hash = calculate_hash(test_file)

print("Original SHA-256:", original_hash)
print("Corrupted SHA-256:", test_hash)

if original_hash != test_hash:
    print("CORRUPTION DETECTED!")
    raise ValueError(
        "INGESTION FAILED: File integrity verification failed."
    )

print("File integrity verified.")
```

```python

# CELL 13 - VERIFY ORIGINAL DATASET

import json
import os

manifest_path = f"{version_path}/manifest.json"

with open(manifest_path, "r", encoding="utf-8") as f:
    saved_manifest = json.load(f)

verified = 0

for item in saved_manifest["files"]:
    filepath = os.path.join(
        version_path, item["filename"]
    )

    assert os.path.exists(filepath), (
        f"Missing file: {item['filename']}"
    )

    assert calculate_hash(filepath) == item["sha256"], (
        f"Hash mismatch: {item['filename']}"
    )

    verified += 1

print("SUCCESS: Original dataset is intact.")
print("Verified files:", verified)
print("Manifest remains valid.")
print("Controlled failure test completed!")
```

```python

# CELL 14 - MEASURE CSV STORAGE SIZE

import os
import time

source_path = "/Volumes/dbw_homecredit/homecredit_bronze/homecredit_uploads"

csv_files = [
    f for f in os.scandir(source_path)
    if f.is_file() and f.name.lower().endswith(".csv")
]

csv_total_bytes = sum(f.stat().st_size for f in csv_files)
csv_size_mb = csv_total_bytes / (1024 ** 2)

print("CSV files:", len(csv_files))
print(f"Total CSV storage: {csv_size_mb:.2f} MB")
```

```python

# CELL 15 - CSV READ PERFORMANCE

csv_path = f"{source_path}/application_train.csv"

start_time = time.perf_counter()

csv_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(csv_path)
)

csv_rows = csv_df.count()
csv_read_seconds = time.perf_counter() - start_time

print("CSV rows:", csv_rows)
print(f"CSV read time: {csv_read_seconds:.2f} seconds")
```

```python

# CELL 16 - CONVERT CSV TO DELTA

delta_path = f"{source_path}/delta_comparison/application_train"

start_time = time.perf_counter()

(
    csv_df.write
    .format("delta")
    .mode("overwrite")
    .save(delta_path)
)

delta_write_seconds = time.perf_counter() - start_time

print("Delta conversion completed!")
print(f"Delta write time: {delta_write_seconds:.2f} seconds")
```

```python

# CELL 17 - DELTA MEASUREMENT

def folder_size(path):
    total = 0
    for item in dbutils.fs.ls(path):
        if item.isDir():
            total += folder_size(item.path)
        else:
            total += item.size
    return total

delta_size_mb = folder_size(delta_path) / (1024 ** 2)

start_time = time.perf_counter()

delta_df = spark.read.format("delta").load(delta_path)
delta_rows = delta_df.count()

delta_read_seconds = time.perf_counter() - start_time

print("Delta rows:", delta_rows)
print(f"Delta storage: {delta_size_mb:.2f} MB")
print(f"Delta read time: {delta_read_seconds:.2f} seconds")
```

```python

# CELL 18 - CSV VS DELTA COMPARISON

csv_training_mb = os.path.getsize(csv_path) / (1024 ** 2)

results = [
    ("CSV", csv_training_mb, csv_read_seconds, csv_rows),
    ("Delta", delta_size_mb, delta_read_seconds, delta_rows)
]

comparison_df = spark.createDataFrame(
    results,
    ["Format", "Storage_MB", "Read_Time_Seconds", "Rows"]
)

display(comparison_df)

assert csv_rows == delta_rows, "Row count mismatch!"

print("CSV vs Delta comparison completed!")
```

```python

# CELL 19 - FINAL RESOURCE MEASUREMENT SUMMARY

from datetime import datetime, timezone
import json
import os

summary = {
    "project": "Home Credit Default Risk",
    "phase": "Phase 1",
    "measurement_date": datetime.now(
        timezone.utc
    ).isoformat(),
    "dataset": "application_train.csv",
    "csv": {
        "size_mb": round(csv_training_mb, 2),
        "read_seconds": round(csv_read_seconds, 2),
        "rows": csv_rows
    },
    "delta": {
        "size_mb": round(delta_size_mb, 2),
        "read_seconds": round(delta_read_seconds, 2),
        "write_seconds": round(delta_write_seconds, 2),
        "rows": delta_rows
    }
}

summary_path = (
    "/Volumes/dbw_homecredit/homecredit_bronze/"
    "homecredit_uploads/phase1_measurements.json"
)

with open(summary_path, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=4)

print(json.dumps(summary, indent=4))
print("Measurement summary saved!")
```
